Cell 1 — Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

RAW_DIR = Path("../data/raw")

pd.set_option("display.max_columns", None)
```

Cell 2 — See all datasets

```python
files = list(RAW_DIR.glob("*.csv"))

for file in files:
    print(f"{file.name:25} {file.stat().st_size / (1024**2):.2f} MB")
```

Cell 3 — Inspect columns without loading everything

```python
for file in files:
    print("\n" + "=" * 60)
    print(file.name)

    try:
        df_sample = pd.read_csv(
            file,
            nrows=5,
            encoding="utf-8"
        )
    except UnicodeDecodeError:
        try:
            df_sample = pd.read_csv(
                file,
                nrows=5,
                encoding="latin-1"
            )
        except Exception as e:
            print("Could not read:", e)
            continue

    print("Columns:", df_sample.columns.tolist())
    display(df_sample.head(2))
```

also tell us which encoding worked

```python
for file in files:
    print("\n" + "=" * 60)
    print(file.name)

    df_sample = None
    encoding_used = None

    for encoding in ["utf-8", "utf-8-sig", "cp1252", "latin-1"]:
        try:
            df_sample = pd.read_csv(
                file,
                nrows=5,
                encoding=encoding
            )
            encoding_used = encoding
            break
        except UnicodeDecodeError:
            continue
        except Exception as e:
            print(f"Other error with {encoding}: {e}")
            break

    if df_sample is None:
        print("Could not read this file")
        continue

    print(f"Encoding: {encoding_used}")
    print("Columns:", df_sample.columns.tolist())
    display(df_sample.head(2))
```
